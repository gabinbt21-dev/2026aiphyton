```python
height1 = float(input('첫 번째 학생의 키(cm): '))
height2 = float(input('두 번째 학생의 키(cm): '))

if height1 > height2:
    diff = height1 - height2
else:
    diff = height2 - height1

print(f'두 학생의 키 차이는 {diff:.1f}cm입니다.')
```

```python
import random

dice1 = random.randint(1, 6)
dice2 = random.randint(1, 6)

print(f'첫 번째 주사위: {dice1}')
print(f'두 번째 주사위: {dice2}')

if dice1 > dice2:
    print('첫 번째 주사위가 큽니다.')
elif dice1 < dice2:
    print('두 번째 주사위가 큽니다.')
else:
    print('두 주사위가 같습니다.')
```

```python
import random

# 1: 가위, 2: 바위, 3: 보
com = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
    print('컴퓨터: 가위')
elif com == 2:
    print('컴퓨터: 바위')
else:
    print('컴퓨터: 보')

if user == com:
    print('비겼습니다.')
elif user == 1 and com == 3:
    print('이겼습니다!')        # 가위 vs 보
elif user == 2 and com == 1:
    print('이겼습니다!')        # 바위 vs 가위
elif user == 3 and com == 2:
    print('이겼습니다!')        # 보 vs 바위
else:
    print('졌습니다.')
```

```python
def asmd(x, y, op):
    if op == '+':
        print(f'{x} + {y} = {x + y}')
    elif op == '-':
        print(f'{x} - {y} = {x - y}')
```

```python
x = int(input('첫번째 수'))
y = int(input('두번째 수'))
op = input('연산자를 입력하시오')

asmd(x, y, op)
```

```python
import random

com = random.randint(1, 100)
count = 0

for i in range(50):

    user = int(input('숫자를 입력하세요: '))
    count = count + 1

    print('컴퓨터가 발생한 숫자:', com)

    if user > com:
        print('입력한 숫자가 더 큽니다.')

    elif user < com:
        print('입력한 숫자가 더 작습니다.')

    else:
        print('정답입니다!')
        print(count, '번 만에 맞췄습니다.')
        break
```
