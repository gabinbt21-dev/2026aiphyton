```python
n = int(input('정수: '))
count = 0

for a in range(1, n + 1):
    if n % a == 0:
        count += 1

print('약수의 개수:', count)
if count == 2:
    print('소수입니다.')
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하시오
```

```python
sum = 0

while True:
    n = int(input('양의 정수 입력 (0: 종료): '))
    if n == 0:
        break
    sum += n

print('합계:', sum)
```
