```python
import numpy as np
import matplotlib.pyplot as plt

months = np.arange(1, 13)
num_cases = [11, 2920, 6636, 750, 615, 1141, 1044, 5473, 3707, 2444, 7324, 26198]
cum_cases = np.cumsum(num_cases)

plt.figure(figsize=(5, 3))
plt.plot(months, num_cases, label='number of cases')
plt.plot(months, cum_cases, label='accumulated number of cases')
plt.xlabel('month')
plt.legend()
plt.show()
```

```python
import numpy as np                      

arr = np.array([10, 20, 30, 40, 50])    
print('arr =', arr)

print('-- (1) 인덱싱 --')
print(arr[0])                           
print(arr[-1])                          

print('-- (2) 슬라이싱 --')
print(arr[1:4])                       
print(arr[:3])                         
print(arr[::2])                        
print(arr[::-1])                       

print('-- (3) 값 바꾸기 --')
arr[0] = 99                             
print(arr)

print('-- (4) 여러 원소 한 번에 선택 --')
print(arr[[0, 2, 4]])                   

print('-- (5) 슬라이스는 원본을 공유한다 --')
part = arr[1:3]                         
part[0] = 0
print('part =', part)
print('arr  =', arr)                    

safe = arr[1:3].copy()                 
safe[0] = -1
print('safe =', safe, ' arr =', arr)   
```

```python
import numpy as np                      

print('-- (1) 데이터 준비 --')
names = np.array(['민수', '지연', '철수', '영희', '준호'])    
subjects = ['국어', '영어', '수학']
scores = np.array([[85, 90, 78],
                   [92, 88, 95],
                   [58, 72, 64],
                   [76, 55, 81],
                   [88, 94, 90]])       
print('shape:', scores.shape)           

print('-- (2) 과목별 분석 (axis=0) --')
print('과목별 평균:', np.round(np.mean(scores, axis=0), 1))   
print('과목별 최고:', np.max(scores, axis=0))
print('과목별 최저:', np.min(scores, axis=0))
print('수학 점수만:', scores[:, 2])     

print('-- (3) 학생별 분석 (axis=1) --')
total = np.sum(scores, axis=1)          
avg = np.round(np.mean(scores, axis=1), 1)   
for i in range(len(names)):
    print(names[i], '총점', total[i], '평균', avg[i])
best = np.argmax(total)                 
print('1등:', names[best], total[best], '점')

print('-- (4) 조건 분석 --')
print('전체 평균:', np.round(np.mean(scores), 1))
print('평균 이상 학생:', names[avg >= np.mean(scores)])      
print('과목별 60점 미만 인원:', np.sum(scores < 60, axis=0))   

print('-- (5) 학점 매기기 --')
grade = np.where(avg >= 90, 'A',
         np.where(avg >= 80, 'B',
          np.where(avg >= 70, 'C', 'D')))    
for n, a, g in zip(names, avg, grade):
    print(n, a, g)
```

```python
import numpy as np                      # [NumPy] 라이브러리 불러오기

np.random.seed(42)                      # [NumPy] seed(): 난수 시작점 고정 -> 매번 같은 결과

print('-- (1) rand : 0 이상 1 미만 실수 --')
print(np.random.rand(3))                # [NumPy] rand(개수): 0~1 사이 실수 난수
print(np.random.rand(2, 2))             # 2행 2열

print('-- (2) randint : 정수 --')
print(np.random.randint(1, 7, size=5))  # [NumPy] randint(시작, 끝, size): 끝(7)은 포함되지 않는다

print('-- (3) normal : 정규분포 --')
heights = np.random.normal(170, 5, 5)   # [NumPy] normal(평균, 표준편차, 개수)
print(np.round(heights, 1))             # [NumPy] round(): 소수 첫째 자리까지

print('-- (4) choice / permutation --')
names = np.array(['가', '나', '다', '라', '마'])
print(np.random.choice(names, 2))       # [NumPy] choice(): 무작위로 2개 뽑기 (중복 가능)
print(np.random.permutation(names))     # [NumPy] permutation(): 순서를 섞은 새 배열

print('-- (5) 주사위 1000번 시뮬레이션 --')
rolls = np.random.randint(1, 7, size=1000)     # 1~6 정수 1000개
for face in range(1, 7):
    count = np.sum(rolls == face)       # [NumPy] 조건에 맞는 개수 세기 (실습 8)
    print(face, '의 눈:', count, '번', round(count / 1000, 3))
print('평균 눈:', np.mean(rolls))        # 이론값 3.5에 가깝다
```

```python
import numpy as np                      # [NumPy] 라이브러리 불러오기

score = np.array([55, 82, 67, 90, 45, 73])
print('score =', score)

print('-- (1) 조건 -> 마스크 --')
mask = score >= 70                      # [NumPy] 비교 결과는 True/False 배열(마스크)
print(mask)

print('-- (2) 불리언 인덱싱 --')
print(score[mask])                      # [NumPy] 마스크가 True인 위치의 값만 선택
print(score[score < 60])                # 조건을 바로 써도 된다

print('-- (3) 개수와 비율 --')
print('70점 이상 인원:', np.sum(score >= 70))     # [NumPy] True는 1로 계산 -> 개수
print('70점 이상 비율:', np.mean(score >= 70))    # 평균 = 비율

print('-- (4) np.where --')
result = np.where(score >= 70, '합격', '불합격')    # [NumPy] where(조건, 참일 때 값, 거짓일 때 값)
print(result)
curved = np.where(score < 60, 60, score)           # 60점 미만은 60점으로 올려준다 (최저 60점)
print(curved)

print('-- (5) 조건 결합 --')
mid = score[(score >= 60) & (score < 80)]          # [NumPy] & : 그리고 (각 조건에 괄호)
print('60~79점:', mid)
edge = score[(score < 50) | (score >= 90)]         # [NumPy] | : 또는
print('50 미만 또는 90 이상:', edge)

print('-- (6) any / all --')
print(np.any(score >= 90))              # [NumPy] any(): 하나라도 True 인가?
print(np.all(score >= 60))              # [NumPy] all(): 모두 True 인가?
```

```python
import numpy as np                      # [NumPy] 라이브러리 불러오기

score = np.array([72, 85, 90, 64, 88])  # 한 학생의 5과목 점수
print('score =', score)

print('-- (1) 기본 통계 --')
print('합계  :', np.sum(score))         # [NumPy] sum(): 합계
print('평균  :', np.mean(score))        # [NumPy] mean(): 평균
print('최대  :', np.max(score))         # [NumPy] max(): 최댓값
print('최소  :', np.min(score))         # [NumPy] min(): 최솟값
print('표준편차:', round(np.std(score), 2))   # [NumPy] std(): 표준편차

print('-- (2) 위치 구하기 --')
print('최댓값 위치:', np.argmax(score))  # [NumPy] argmax(): 최댓값이 있는 인덱스
print('최솟값 위치:', np.argmin(score))  # [NumPy] argmin(): 최솟값이 있는 인덱스

print('-- (3) 2차원 배열과 axis --')
table = np.array([[80, 70, 90],          # 학생 A
                  [60, 85, 75],          # 학생 B
                  [95, 90, 100],         # 학생 C
                  [50, 65, 70]])         # 학생 D  (열: 국어, 영어, 수학)
print(table.shape)                       # (4, 3) = 4명 x 3과목
print('전체 평균  :', np.mean(table))
print('과목별 평균:', np.mean(table, axis=0))   # [NumPy] axis=0: 세로 방향 -> 열(과목)마다 계산
print('학생별 평균:', np.mean(table, axis=1))   # [NumPy] axis=1: 가로 방향 -> 행(학생)마다 계산
print('학생별 합계:', np.sum(table, axis=1))
print('과목별 최고:', np.max(table, axis=0))

print('-- (4) 정렬과 누적합 --')
a = np.array([5, 2, 9, 1])
print(np.sort(a))                        # [NumPy] sort(): 오름차순 정렬 (원본은 그대로)
print(np.cumsum(a))                      # [NumPy] cumsum(): 누적합 -> 5, 7, 16, 17
```
